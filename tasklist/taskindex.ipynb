```python
import json
import pandas as pd
from IPython.display import display, HTML, FileLink
from pathlib import Path

with open("habitable_index.json") as f:
    data = json.load(f)

parts = data["parts"]

print("📂 章一覧（クリックでファイルを開く）")
print("-" * 50)

for title, filename in parts.items():
    if Path(filename).exists():
        link = FileLink(filename)
        print(f"📁 {title}")
        display(link)
        print()
    else:
        print(f"⚠️ {title}: {filename} が見つかりません")
```
