```python
# ============================================================
# main.ipynb - 超シンプルダッシュボード（警告ゼロ・完全フリースタイル）
# 
# Version: 3.0 (2026-08-12)
#   - 警告を完全に削除（何も表示しない）
#   - 変換を完全に削除（status をそのまま表示）
#   - 日付自動除去をオフ（完全フリースタイル）
#   - ファイルリンク機能を維持
#   - 固まらない設計（clear_output 方式）
# ============================================================

import json
from pathlib import Path
from IPython.display import display, clear_output, HTML
import ipywidgets as widgets

CONFIG_PATH = "/home/iwamura/ドキュメント/coupling/working/assembly_config_auto.json"
STATUS_LIST = ["⬜ 未入力", "🔄 実行中", "✅ 完了", "❌ エラー", "❓ 不明"]

# ============================================================
# フリースタイル：何も変換しない！何も警告しない！
# ============================================================
def normalize_status(status):
    """完全フリースタイル：何も変換せず、そのまま返す"""
    if status is None or not isinstance(status, str):
        return "❓ 不明"
    return status  # そのまま！

# ============================================================
# JSON操作
# ============================================================
def load_config():
    with open(CONFIG_PATH, 'r', encoding='utf-8') as f:
        return json.load(f)

def save_config(config):
    with open(CONFIG_PATH, 'w', encoding='utf-8') as f:
        json.dump(config, f, indent=2, ensure_ascii=False, sort_keys=True)

def set_status(part_name, new_status):
    config = load_config()
    if part_name in config.get("parts", {}):
        config["parts"][part_name]["status"] = new_status
        save_config(config)
    build_dashboard()

# ============================================================
# ダッシュボード（警告ゼロ）
# ============================================================
def build_dashboard():
    clear_output(wait=True)
    config = load_config()
    
    print("=" * 70)
    print(f"📁 {config['project']}  |  🕒 {config.get('last_scan', '')[:16]}")
    print(f"📌 main.ipynb v3.0（警告ゼロ・完全フリースタイル）")
    print("💡 状態を選んで [適用] → 自動更新")
    print("=" * 70)
    
    for name, part in config["parts"].items():
        status = part.get("status", "❓ 不明")
        display_name = part.get("display_name", name)
        full_path = part.get("full_path", "")
        
        print(f"\n📦 {display_name}")
        
        if full_path and Path(full_path).exists():
            jupyter_path = full_path.replace("/home/iwamura/", "")
            display(HTML(f'    🔗 <a href="/lab/tree/{jupyter_path}" target="_top">ファイルを開く</a>'))
        else:
            print("    🔗 ファイルなし")
        
        # ★ プルダウンには現在の値をそのまま表示
        options = STATUS_LIST.copy()
        if status not in options:
            options.append(status)
        
        dropdown = widgets.Dropdown(
            options=options,
            value=status if status in options else options[0],
            description="変更:",
            layout=widgets.Layout(width='280px')
        )
        
        apply_btn = widgets.Button(
            description="適用",
            layout=widgets.Layout(width='60px')
        )
        
        def create_handler(part_name_param, dropdown_param):
            def handler(b):
                set_status(part_name_param, dropdown_param.value)
            return handler
        
        apply_btn.on_click(create_handler(name, dropdown))
        display(widgets.HBox([dropdown, apply_btn]))
        
        deps = part.get("depends_on", [])
        if deps:
            print(f"    📦 依存: {', '.join(deps)}")

# ============================================================
# 実行
# ============================================================
clear_output()
build_dashboard()
```
